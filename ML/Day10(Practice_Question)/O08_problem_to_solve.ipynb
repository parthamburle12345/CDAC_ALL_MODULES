```python
import numpy as np
import pandas as pd
```

```python
r_cols = ['user_id', 'movie_id', 'rating']
ratings = pd.read_csv("./assets/u.data",sep="\t",usecols=range(3), names=r_cols,encoding='ISO-8859-1')
print(ratings.head())
# print("Shape of ratings:")
# print(ratings.shape)
print("=" * 60)
m_cols = ['movie_id', 'title']
movies = pd.read_csv("./assets/u.item",sep="|",usecols=range(2), names=m_cols,encoding='ISO-8859-1')
print(movies.head())
print("Shape of movies:")
print(movies.shape)
print("=" * 60)
ratings = pd.merge(movies, ratings)
print(ratings.head())
# print("Shape of merged ratings:")
# print(ratings.shape)
print("=" * 60)
```

```python
print("=" * 60)
moviesRatings = ratings.pivot_table(index=['user_id'], columns=['title'], values='rating')
print(moviesRatings.head())
```

```python
myMovie = moviesRatings['Star Wars (1977)']
print(myMovie.head())
```

```python
similarMovies = moviesRatings.corrwith(myMovie)
similarMovies = similarMovies.dropna()
df = pd.DataFrame(similarMovies, columns=['similarity'])
print(df.head())
```

```python
print(df.sort_values(by='similarity', ascending=False).head(10))
```

```python
import numpy as np
movieStats = ratings.groupby('title').agg({'rating': ['count', 'mean']})
print(movieStats.head())
```

```python
popularMovies = movieStats['rating']['count'] >= 100
print(movieStats[popularMovies].sort_values([('rating', 'mean')], ascending=False)[:15])
```

```python
# Filter recommendations: Join similarity scores with rating counts
df = df.join(movieStats['rating']['count'])
df = df.join(movieStats['rating']['mean'])

# Recommend movies with >= 100 ratings sorted by similarity to Star Wars (1977)
star_wars_recommendations = df[df['count'] >= 100].sort_values(by='similarity', ascending=False)
print("Top 10 Movie Recommendations for Fans of Star Wars (1977):")
print(star_wars_recommendations.head(10))
```
