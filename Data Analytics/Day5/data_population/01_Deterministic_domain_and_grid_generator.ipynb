### * Setup Imports & plotting congifuration


```python
# Deterministic domain and grid generator
import numpy as np
import matplotlib.pyplot as plt

#apply visualization style
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
#seaborn-v0_8-whitegrid:- a clean white grid style inspired by seaborn's whitegrid style, with a focus on clarity and simplicity.

plt.rcParams["figure.dpi"] = 100
#rc means "run commands" or "runtime configuration". It is a module in Matplotlib that allows you to customize various aspects of the library's behavior and appearance. The rcParams object is a dictionary-like object that contains all the default settings for Matplotlib, such as figure size, font size, line width, color maps, and more. By modifying the values in rcParams, you can change the default behavior of Matplotlib to suit your needs.
```

### 1. Linear spaceing ('np.linespace')

```python
#retstep=True: This argument is used in the np.linspace() function to indicate that you want to return the step size between the generated values.
# When retstep=True, the function will return a tuple containing two elements: the array of evenly spaced values and the step size between those values.
x_linear,step_size = np.linspace(start=0, stop=10, num=15, retstep=True)
print("Linear space values:", x_linear)
print("Step size:", step_size)
plt.figure(figsize=(8, 4))

#scatter plot parameter details:
#s=60: This sets the size of the points in the scatter plot.
#zorder=3: This sets the z-order of the points in the scatter plot, ensuring they are plotted on top of other elements.
#label='Linear Space Points': This sets the label for the points in the scatter plot, which will be used in the legend.
#alpha=0.5: This sets the transparency level of the vertical lines, making them semi-transparent.(0.0 is fully transparent, 1.0 is fully opaque)
#linestyle='--': This sets the line style of the vertical lines to be dashed.


plt.scatter(x_linear, np.zeros_like(x_linear), color='blue', s=60, zorder=3, label='Linear Space (h={step_size})'.format(step_size=step_size))

plt.vlines(x_linear, ymin=-0.5, ymax=0.5, color='blue', alpha=0.5, zorder=2, linestyle='--')

plt.title('1D uniform linear space with vertical lines', fontsize=14, fontweight='bold')
plt.xlabel('values', fontsize=14)
plt.yticks([])  #hide y-axis ticks since the y-axis is not relevant for this plot
plt.ylim(-0.5, 0.5)  #set y-axis limits to ensure the vertical lines are fully visible
plt.legend(loc='upper right', fontsize=10)
plt.tight_layout()  #adjust the padding between and around subplots to minimize overlap and ensure that all elements of the plot are visible and well-spaced    
plt.show()  
plt
```

### 2. log spacing

```python
# generate 7 points from 10^-5(0.00001) to 10^1(10.0) in logarithmic space
learning_rate = np.logspace(start=-5, stop=1, num=7,base=10)
print("Logarithmic space array:", learning_rate)
print("Generates hyperparameter learning_rate")
# for i , lr in enumerate(learning_rate):
#     print(f"learning_rate[{i}] = {lr:.5f}")
# print()
print("learning_rate", [f"{lr:.5f}" for lr in learning_rate])

#subplot paramaeters:
#2,1: This indicates that the figure will have 2 rows and 1 column of subplots.
#figsize=(12, 4): This sets the size of the entire figure to be 12 inches wide and 4 inches tall.
fig,(ax1,ax2) = plt.subplots(1,2,figsize=(12,4))


#top plot: logarithmic scale visualization
ax1.scatter(learning_rate, np.zeros_like(learning_rate), color='green', s=60, zorder=3,label='Logarithmic Space (learning_rate)')
# ax1.vlines(learning_rate, ymin=-0.5, ymax=0.5)
ax1.set_xscale('log')  #set x-axis to logarithmic scale
ax1.set_title('Logarithmic space for learning_rate', fontsize=14, fontweight='bold')
ax1.set_xlabel('learning_rate', fontsize=14)
ax1.set_yticks([])  #hide y-axis ticks since the y-axis is not relevant for this plot

#bottom plot: linear scale visualization
ax2.scatter(learning_rate, np.zeros_like(learning_rate), color='orange', s=60, zorder=3,label='linear Space (learning_rate)')
ax2.set_title('Logarithmic space for learning_rate (linear scale)', fontsize=14, fontweight='bold')
ax2.set_xlabel('learning_rate', fontsize=14)
ax2.set_ylabel('Frequency', fontsize=14)

```

```python
geom_grid = np.geomspace(start=1.0, stop=1000, num=6)
print("Geometric space array:", geom_grid)
print("common multiplicative ratio r = geom_grid[1]/geom_grid[0] = ", f"{geom_grid[1]/geom_grid[0]:.4f}")

#graphical representation of geometric space

plt.figure(figsize=(9, 3.5))

#plot formatting parameters details:
#range(len(geom_grid)): This generates a sequence of integers from 0 to the length
#0-- > len(geom_grid)-1, which will be used as the x-coordinates for the plot.
#geom_grid: This is the array of geometric space values that will be plotted on the y
#marker='o': This specifies that the data points should be marked with circles on the plot.

plt.plot(range(len(geom_grid)), geom_grid, marker='o', color='purple', linewidth=2, markersize=8, label='np.geomspace(1 to 1000)')
plt.title('Geometric progression growth curve', fontsize=14, fontweight='bold')
plt.xlabel('Index', fontsize=14)
plt.ylabel('Generated Value', fontsize=14)
plt.yscale('log')  #set y-axis to logarithmic scale
plt.grid(True, which='both', linestyle='--', alpha=0.5)  #enable grid lines for both major and minor ticks with dashed lines
plt.legend()
plt.show()
```

### 4. stemplot

```python
# stem plot for time steps


time_steps = np.arange(start=0, stop=5, step=0.5)
print("Time steps array:", time_steps)
print("common additive difference d = time_steps[1]-time_steps[0] = ", f"{time_steps[1]-time_steps[0]:.4f}")
print("total number of time steps = ", len(time_steps))

#what is stem plot?
#stemm plot is a type of plot that displays data points as vertical lines (stems) extending from a baseline,
#  with markers at the top of each stem to indicate the data values. It is useful for visualizing discrete data points
#  and their distribution along an axis.


# graphical representation of time steps
plt.figure(figsize=(9, 3.5))
#plot formatting parameters details:
#linefmt = c4olor='red', linestyle='--', linewidth=1.5, alpha=0.7
#color cycle definations c1 = 'blue', c2 = 'orange', c3 = 'green', c4 = 'red', c5 = 'purple', c6 = 'brown', c7 = 'pink', c8 = 'gray', c9 = 'olive', c10 = 'cyan'
#'markerfmt' = 'o', markersize=8, markerfacecolor='red', markeredgecolor='black', markeredgewidth=1.5, alpha=0.9
#'basefmt' = 'black', linewidth=1.5, alpha=0.7 specifies the format of the baseline of the stem plot, which is the horizontal line that connects the stems to the x-axis. It sets the color of the baseline to black, the linewidth to 1.5, and the alpha (transparency) to 0.7.


plt.stem(time_steps, np.ones_like(time_steps), linefmt='r--', markerfmt='ro', basefmt='k-')
plt.title('Time steps representation using stem plot', fontsize=14, fontweight='bold')
plt.xlabel('Time steps', fontsize=14)
plt.yticks([])  #hide y-axis ticks since the y-axis is not relevant for this plot
plt.ylim(-0.5, 1.5)  #set y-axis limits to ensure the stems are fully visible
plt.tight_layout()  #adjust the padding between and around subplots to minimize overlap and ensure that all elements of the plot are visible and well-spaced
plt.show()
```

### 5.coordinate grids(np.meshgrid)

```python
x_domain = np.linspace(-2, 2, 20)
y_domain = np.linspace(-2, 2, 20)

print("X domain:", x_domain)
print("*" * 80)
print("Y domain:", y_domain)
print("*" * 80)

#construct 2d coordinate matrices
X, Y = np.meshgrid(x_domain, y_domain)
#x is the row repeated matrix for y_domain and y is the column repeated matrix for x_domain
#y is transposed compare to x 


#evaluate 2d parabola function z = x^2 + y^2
Z = X**2 + Y**2

print("Grid coordinates output:")
print(f"Matrix {X.shape})")
print(f"Matrix {Y.shape})")
print(f"Matrix {Z.shape})")
print("*" * 80)
print("sample X matrix slice ( first 3 x 3 blocks):\n ", X[:3, :3])
print("*" * 80)
print("sample Y matrix slice ( first 3 x 3 blocks):\n ", Y[:3, :3])
print("*" * 80)
print("sample Z matrix slice ( first 3 x 3 blocks):\n ", Z[:3, :3])
print()

#graphical visualization of 2d parabola function z = x^2 + y^2
fig,(ax1,ax2) = plt.subplots(1,2,figsize=(12,4))
# s is the size of the scatter plot marker 
# subplot 1: scatter plot of the 2D coordinate grid mesh
ax1.scatter( X, Y, c="blue", marker='o', s=15)
ax1.set_title('2D coordinate grid mesh (np.meshgrid)', fontsize=14, fontweight='bold')
ax1.set_xlabel('X-axis', fontsize=14)   
ax1.set_ylabel('Y-axis', fontsize=14)
ax1.set_aspect('equal', adjustable='box')  #set equal aspect ratio

#subplot 2: contour plot of the 2D coordinate grid z = x^2 + y^2

#parameters details:
#cmap = colormap used for the contour plot. 'viridis' is a perceptually uniform colormap that is often used for scientific visualization.
#
counter = ax2.contourf(X, Y, Z, levels=15, cmap='viridis')  #filled contour plot with 20 levels and viridis colormap
#countour and counterf are used to create contour plots in Matplotlib. A contour plot is a graphical representation of a 3D surface on a 2D plane, where lines or filled regions represent constant values of a function. In this case, the function is z = x^2 + y^2, and the contour plot shows how the function's value changes across the 2D grid defined by X and Y.
fig.colorbar(counter, ax=ax2 , label='Z-axis')  #add colorbar to the contour plot
ax2.set_title('Contour plot of 2D parabola function z = x^2 + y^2', fontsize=14, fontweight='bold')







# ============================================================
# 3D visualization of parabola function Z = X² + Y²
# ============================================================

fig = plt.figure(figsize=(10, 7))

# Create 3D axis
ax = fig.add_subplot(111, projection='3d')

# Create 3D surface
surface = ax.plot_surface(
    X,
    Y,
    Z,
    cmap='viridis'
)

# Labels
ax.set_xlabel('X-axis', fontsize=12)
ax.set_ylabel('Y-axis', fontsize=12)
ax.set_zlabel('Z-axis', fontsize=12)

# Title
ax.set_title('3D Paraboloid: Z = X² + Y²', fontsize=14, fontweight='bold')

# Add color bar
fig.colorbar(surface, ax=ax, shrink=0.6, label='Z value')
plt.show()

```
